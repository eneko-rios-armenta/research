## Speed vs Stimulation

### graphing

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter

def plot_speed_vs_time_with_stimulation(base_folder, threshold=500):
    """
    Analyze and plot the speed of larvae with time along with stimulation.
    Smooths data, calculates speed, and truncates based on stimulation threshold.

    Parameters:
        base_folder (str): Path to the base folder containing subfolders with GUI-corrected files.
        threshold (float): The stimulation threshold below which data is excluded.

    Returns:
        None
    """
    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv") and not f.startswith("aligned_")
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Midpoint', 'Y-Midpoint', 'Time', 'stimulation']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Smooth the data using Savitzky-Golay filter
        window_length = 31  # Must be odd and less than the length of the data
        polyorder = 3       # Polynomial order for the filter
        data['X-Midpoint'] = savgol_filter(data['X-Midpoint'], window_length, polyorder)
        data['Y-Midpoint'] = savgol_filter(data['Y-Midpoint'], window_length, polyorder)

        # Calculate the speed
        x_midpoint = data['X-Midpoint'].values
        y_midpoint = data['Y-Midpoint'].values
        time = data['Time'].values
        stimulation = data['stimulation'].values

        # Calculate the differences in positions and time
        dx = np.diff(x_midpoint)
        dy = np.diff(y_midpoint)
        dt = np.diff(time)

        # Avoid division by zero
        dt[dt == 0] = np.nan

        # Calculate speed (pixels per unit time)
        speed = np.sqrt(dx**2 + dy**2) / dt
        speed = np.clip(speed, 0, np.nanpercentile(speed, 95))  # Remove extreme outliers

        # Apply moving average to smooth the speed and stimulation data
        speed_smoothed = pd.Series(speed).rolling(window=window_length, center=True).mean().to_numpy()
        stimulation_smoothed = pd.Series(stimulation).rolling(window=window_length, center=True).mean().to_numpy()

        # Identify the last index where stimulation crosses the threshold
        below_threshold = stimulation_smoothed < threshold
        if np.any(below_threshold):
            last_below_index = np.argmax(below_threshold)  # First index where stimulation falls below the threshold
        else:
            last_below_index = len(stimulation_smoothed)  # Use the entire dataset if no threshold is crossed

        # Truncate the data
        time = time[:last_below_index]
        speed_smoothed = speed_smoothed[:last_below_index]
        stimulation_smoothed = stimulation_smoothed[:last_below_index]

        # Plot speed vs time with stimulation
        fig, ax1 = plt.subplots(figsize=(10, 6))

        # Plot speed on the primary y-axis
        if (len(time) == speed_smoothed.shape[0]) and (len(time) == stimulation_smoothed.shape[0]):
            ax1.plot(time[0:], speed_smoothed, label="Speed", color="blue")
        else:
            ax1.plot(time[1:], speed_smoothed, label="Speed", color="blue")

        ax1.set_xlabel("Time (s)")
        ax1.set_ylabel("Speed (pixels/s)", color="blue")
        ax1.tick_params(axis='y', labelcolor="blue")
        ax1.grid(True)

        # Plot stimulation on the secondary y-axis
        ax2 = ax1.twinx()
        ax2.plot(time, stimulation_smoothed, label="Stimulation", color="orange", alpha=0.7)
        ax2.set_ylabel("Stimulation", color="orange")
        ax2.tick_params(axis='y', labelcolor="orange")

        # Add title and legend
        plt.title(f"Speed and Stimulation vs Time for {folder}")
        fig.tight_layout()
        plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/250421/gaussian'  # Replace with your base folder path
plot_speed_vs_time_with_stimulation(base_folder, threshold=500)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter

def plot_speed_vs_time_with_stimulation_and_correlation_avgplume_experiments(base_folder, threshold=500):
    """
    Analyze and plot the speed of larvae with time along with stimulation.
    Smooths data, calculates speed, truncates based on stimulation threshold, and calculates correlation.

    Parameters:
        base_folder (str): Path to the base folder containing subfolders with GUI-corrected files.
        threshold (float): The stimulation threshold below which data is excluded.

    Returns:
        None
    """
    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv") and (f.startswith("aligned_aligned_") or f.startswith("aligned_"))
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Midpoint', 'Y-Midpoint', 'Time', 'stimulation']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Smooth the data using Savitzky-Golay filter
        window_length = 31  # Must be odd and less than the length of the data
        polyorder = 3       # Polynomial order for the filter
        data['X-Midpoint'] = savgol_filter(data['X-Midpoint'], window_length, polyorder)
        data['Y-Midpoint'] = savgol_filter(data['Y-Midpoint'], window_length, polyorder)

        # Calculate the speed
        x_midpoint = data['X-Midpoint'].values
        y_midpoint = data['Y-Midpoint'].values
        time = data['Time'].values
        stimulation = data['stimulation'].values

        # Calculate the differences in positions and time
        dx = np.diff(x_midpoint)
        dy = np.diff(y_midpoint)
        dt = np.diff(time)

        # Avoid division by zero
        dt[dt == 0] = np.nan

        # Calculate speed (pixels per unit time)
        speed = np.sqrt(dx**2 + dy**2) / dt
        speed = np.clip(speed, 0, np.nanpercentile(speed, 95))  # Remove extreme outliers

        # Apply moving average to smooth the speed and stimulation data
        speed_smoothed = pd.Series(speed).rolling(window=window_length, center=True).mean().to_numpy()
        stimulation_smoothed = pd.Series(stimulation).rolling(window=window_length, center=True).mean().to_numpy()

        # Ensure speed_smoothed and stimulation_smoothed are the same length
        min_length = min(len(speed_smoothed), len(stimulation_smoothed))
        speed_smoothed = speed_smoothed[:min_length]
        stimulation_smoothed = stimulation_smoothed[:min_length]

        # Identify the last index where stimulation crosses the threshold
        below_threshold = stimulation_smoothed < threshold
        if np.any(below_threshold):
            last_below_index = np.argmax(below_threshold)  # First index where stimulation falls below the threshold
        else:
            last_below_index = len(stimulation_smoothed)  # Use the entire dataset if no threshold is crossed

        # Truncate the data
        time = time[:last_below_index]
        speed_smoothed = speed_smoothed[:last_below_index]
        stimulation_smoothed = stimulation_smoothed[:last_below_index]

        # Calculate and print the correlation
        valid_indices = ~np.isnan(speed_smoothed) & ~np.isnan(stimulation_smoothed)
        if np.sum(valid_indices) > 1:  # Ensure there are enough valid points
            correlation = np.corrcoef(speed_smoothed[valid_indices], stimulation_smoothed[valid_indices])[0, 1]
            print(f"Correlation between speed and stimulation for {folder}: {correlation:.2f}")
        else:
            print(f"Not enough valid data points for correlation in {folder}. Skipping correlation calculation.")

        # Plot speed vs time with stimulation
        fig, ax1 = plt.subplots(figsize=(10, 6))

        # Plot speed on the primary y-axis
        if (len(time) == speed_smoothed.shape[0]) and (len(time) == stimulation_smoothed.shape[0]):
            ax1.plot(time[0:], speed_smoothed, label="Speed", color="blue")
        else:
            ax1.plot(time[1:], speed_smoothed, label="Speed", color="blue")

        ax1.set_xlabel("Time (s)")
        ax1.set_ylabel("Speed (pixels/s)", color="blue")
        ax1.tick_params(axis='y', labelcolor="blue")
        ax1.grid(True)

        # Plot stimulation on the secondary y-axis
        ax2 = ax1.twinx()
        ax2.plot(time, stimulation_smoothed, label="Stimulation", color="orange", alpha=0.7)
        ax2.set_ylabel("Stimulation", color="orange")
        ax2.tick_params(axis='y', labelcolor="orange")

        # Add title and legend
        plt.title(f"Speed and Stimulation vs Time for {folder}")
        fig.tight_layout()
        plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/250501/average_plume recenter' # Replace with your base folder path
plot_speed_vs_time_with_stimulation_and_correlation_avgplume_experiments(base_folder, threshold=500)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.ndimage import gaussian_filter
from dtaidistance import dtw
from scipy.spatial.distance import euclidean

def plot_speed_vs_time_with_stimulation_dtw(base_folder):
    """
    Analyze and plot the speed of larvae with time along with stimulation.
    Uses DTW to calculate the lag between speed and stimulation.

    Parameters:
        base_folder (str): Path to the base folder containing subfolders with GUI-corrected files.

    Returns:
        None
    """
    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv")
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Midpoint', 'Y-Midpoint', 'Time', 'stimulation']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Smooth the data using Savitzky-Golay filter
        window_length = 31  # Must be odd and less than the length of the data
        polyorder = 3       # Polynomial order for the filter
        data['X-Midpoint'] = savgol_filter(data['X-Midpoint'], window_length, polyorder)
        data['Y-Midpoint'] = savgol_filter(data['Y-Midpoint'], window_length, polyorder)

        # Calculate the speed
        x_midpoint = data['X-Midpoint'].values
        y_midpoint = data['Y-Midpoint'].values
        time = data['Time'].values
        stimulation = data['stimulation'].values

        # Calculate the differences in positions and time
        dx = np.diff(x_midpoint)
        dy = np.diff(y_midpoint)
        dt = np.diff(time)

        # Avoid division by zero
        dt[dt == 0] = np.nan

        # Calculate speed (pixels per unit time)
        speed = np.sqrt(dx**2 + dy**2) / dt
        speed = np.clip(speed, 0, np.nanpercentile(speed, 95))  # Remove extreme outliers

        # Apply moving average to smooth the speed and stimulation data
        speed_smoothed = pd.Series(speed).rolling(window=window_length, center=True).mean().to_numpy()
        stimulation_smoothed = pd.Series(stimulation).rolling(window=window_length, center=True).mean().to_numpy()

        # Apply gradient to stimulation to get stimulation gradient
        stimulation_smoothed = gaussian_filter(np.gradient(stimulation_smoothed), sigma=100)

        # Apply Gaussian filter to smooth the speed
        speed_smoothed = gaussian_filter(speed_smoothed, sigma=100)

        # Ensure speed_smoothed and stimulation_smoothed are the same length
        min_length = min(len(speed_smoothed), len(stimulation_smoothed))
        speed_smoothed = speed_smoothed[:min_length]
        stimulation_smoothed = stimulation_smoothed[:min_length]

        # Remove NaN values and ensure 1-D arrays
        valid_indices = ~np.isnan(speed_smoothed) & ~np.isnan(stimulation_smoothed)
        speed_smoothed = speed_smoothed[valid_indices].flatten()  # Ensure 1-D
        stimulation_smoothed = stimulation_smoothed[valid_indices].flatten()  # Ensure 1-D

        # Use DTW to calculate the lag
        #return speed_smoothed.shape, stimulation_smoothed.shape
        # Ensure speed_smoothed and stimulation_smoothed are 1-D arrays
        speed_smoothed = np.ravel(speed_smoothed)
        stimulation_smoothed = np.ravel(stimulation_smoothed)

        # Use DTW to calculate the lag
        speed_smoothed.reshape(-1)
        path = dtw.warping_path(speed_smoothed, stimulation_smoothed)
        lag = path[0][0] - path[0][1]  # Calculate lag based on alignment path

        print(f"Folder: {folder}")
        print(f"DTW Lag: {lag} frames")
        #print(f"DTW Distance: {distance:.3f}")

        # Adjust stimulation signal based on the lag
        if lag > 0:
            stimulation_smoothed = stimulation_smoothed[lag:]
            speed_smoothed = speed_smoothed[:len(stimulation_smoothed)]
            time = time[:len(stimulation_smoothed)]
        elif lag < 0:
            speed_smoothed = speed_smoothed[-lag:]
            stimulation_smoothed = stimulation_smoothed[:len(speed_smoothed)]
            time = time[:len(speed_smoothed)]

        # Plot speed vs time with stimulation
        fig, ax1 = plt.subplots(figsize=(10, 6))

        # Plot speed on the primary y-axis
        # Ensure speed_smoothed and time are the same size
        min_length = min(len(speed_smoothed), len(time))
        speed_smoothed = speed_smoothed[:min_length]
        time = time[:min_length]
        
        ax1.plot(time, speed_smoothed, label="Speed", color="blue")
        ax1.set_xlabel("Time (s)")
        ax1.set_ylabel("Speed (pixels/s)", color="blue")
        ax1.tick_params(axis='y', labelcolor="blue")
        ax1.grid(True)

        # Plot stimulation on the secondary y-axis
        ax2 = ax1.twinx()
        ax2.plot(time, stimulation_smoothed, label="Stimulation", color="orange", alpha=0.7)
        ax2.set_ylabel("Stimulation", color="orange")
        ax2.tick_params(axis='y', labelcolor="orange")

        # Add title and legend
        plt.title(f"Speed and Stimulation vs Time for {folder}\nDTW Lag: {lag} frames")
        fig.tight_layout()
        plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/250422/gaussian'  # Replace with your base folder path
plot_speed_vs_time_with_stimulation_dtw(base_folder)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter
from scipy.ndimage import gaussian_filter

def plot_speed_vs_time_with_stimulation_gradient_and_correlation(base_folder):
    """
    Analyze and plot the speed of larvae with time along with stimulation.
    Smooths data, calculates speed, shifts speed by 118 frames to align with stimulation,
    and calculates correlation.

    Parameters:
        base_folder (str): Path to the base folder containing subfolders with GUI-corrected files.

    Returns:
        None
    """
    # Define the frame shift (118 frames)
    frame_shift = 118
    
    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv") # and (f.startswith("aligned_aligned_") or f.startswith("aligned_"))  # uncomment for experiments that rotate the arena
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Midpoint', 'Y-Midpoint', 'Time', 'stimulation']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Smooth the data using Savitzky-Golay filter
        window_length = 31  # Must be odd and less than the length of the data
        polyorder = 3       # Polynomial order for the filter
        data['X-Midpoint'] = savgol_filter(data['X-Midpoint'], window_length, polyorder)
        data['Y-Midpoint'] = savgol_filter(data['Y-Midpoint'], window_length, polyorder)

        # Calculate the speed
        x_midpoint = data['X-Midpoint'].values
        y_midpoint = data['Y-Midpoint'].values
        time = data['Time'].values
        stimulation = data['stimulation'].values

        # Calculate the differences in positions and time
        dx = np.diff(x_midpoint)
        dy = np.diff(y_midpoint)
        dt = np.diff(time)

        # Avoid division by zero
        dt[dt == 0] = np.nan

        # Calculate speed (pixels per unit time)
        speed = np.sqrt(dx**2 + dy**2) / dt
        speed = np.clip(speed, 0, np.nanpercentile(speed, 95))  # Remove extreme outliers

        # Apply moving average to smooth the speed and stimulation data
        speed_smoothed = pd.Series(speed).rolling(window=window_length, center=True).mean().to_numpy()
        stimulation_smoothed = pd.Series(stimulation).rolling(window=window_length, center=True).mean().to_numpy()

        # Apply gradient to stimulation to get stimulation gradient
        stimulation_smoothed = gaussian_filter(np.gradient(stimulation_smoothed), sigma=100)

        # Apply Gaussian filter to smooth the speed
        speed_smoothed = gaussian_filter(speed_smoothed, sigma=100)

        # SHIFT THE SPEED DATA BY 118 FRAMES TO ALIGN WITH STIMULATION
        # Shift speed_smoothed forward by 118 frames (remove first 118 frames from speed)
        if len(speed_smoothed) > frame_shift:
            speed_smoothed_shifted = speed_smoothed[frame_shift:]
            # Remove last 118 frames from stimulation to match length
            stimulation_smoothed_aligned = stimulation_smoothed[:-frame_shift]
            time_aligned = time[:-frame_shift]  # Also align time
        else:
            print(f"Not enough frames to shift by {frame_shift} in {folder}. Using original data.")
            speed_smoothed_shifted = speed_smoothed
            stimulation_smoothed_aligned = stimulation_smoothed
            time_aligned = time

        # Ensure aligned arrays are the same length
        min_length = min(len(speed_smoothed_shifted), len(stimulation_smoothed_aligned))
        speed_smoothed_aligned = speed_smoothed_shifted[:min_length]
        stimulation_smoothed_aligned = stimulation_smoothed_aligned[:min_length]
        time_aligned = time_aligned[:min_length]

        # Calculate and print the correlation with aligned data
        valid_indices = ~np.isnan(speed_smoothed_aligned) & ~np.isnan(stimulation_smoothed_aligned)
        if np.sum(valid_indices) > 1:  # Ensure there are enough valid points
            correlation = np.corrcoef(speed_smoothed_aligned[valid_indices], stimulation_smoothed_aligned[valid_indices])[0, 1]
            print(f"Correlation between speed (shifted by {frame_shift} frames) and stimulation for {folder}: {correlation:.2f}")
        else:
            print(f"Not enough valid data points for correlation in {folder}. Skipping correlation calculation.")

        # Plot speed vs time with stimulation (using aligned data)
        fig, ax1 = plt.subplots(figsize=(10, 6))
        
        # Plot speed on the primary y-axis
        ax1.plot(time_aligned, speed_smoothed_aligned, label=f"Speed (shifted by {frame_shift} frames)", color="blue")
        ax1.set_xlabel("Time (s)")
        ax1.set_ylabel("Speed (pixels/s)", color="blue")
        ax1.tick_params(axis='y', labelcolor="blue")
        ax1.grid(True)

        # Plot stimulation on the secondary y-axis
        ax2 = ax1.twinx()
        ax2.plot(time_aligned, stimulation_smoothed_aligned, label="Stimulation", color="orange", alpha=0.7)
        ax2.set_ylabel("Stimulation", color="orange")
        ax2.tick_params(axis='y', labelcolor="orange")

        # Add title and legend
        plt.title(f"Speed (shifted by {frame_shift} frames) and Stimulation vs Time for {folder}")
        fig.tight_layout()
        plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/250422/gaussian' # Replace with your base folder path
plot_speed_vs_time_with_stimulation_gradient_and_correlation(base_folder)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import savgol_filter

def compare_baseline_speeds_with_boxplots(folder1, folder2):
    """
    Compare the speed data of all experiments in two baseline test folders using box plots.

    Parameters:
        folder1 (str): Path to the first baseline test folder.
        folder2 (str): Path to the second baseline test folder.

    Returns:
        None
    """
    def calculate_speeds(base_folder):
        """
        Calculate the speed data for all experiments in a folder.

        Parameters:
            base_folder (str): Path to the folder containing subfolders with GUI-corrected files.

        Returns:
            list: A list of average speeds for each experiment in the folder.
        """
        all_speeds = []

        for folder in os.listdir(base_folder):
            folder_path = os.path.join(base_folder, folder)
            if not os.path.isdir(folder_path):
                continue

            # Find the GUI-corrected file in the folder
            gui_corrected_files = [
                f for f in os.listdir(folder_path)
                if f.endswith("GUIcorrected.csv") and not f.startswith("aligned_")
            ]
            if not gui_corrected_files:
                print(f"No GUI-corrected file found in {folder_path}. Skipping.")
                continue

            gui_corrected_file = gui_corrected_files[0]
            gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

            # Read the GUI-corrected file
            data = pd.read_csv(gui_corrected_path)

            # Ensure required columns exist
            if 'X-Midpoint' not in data.columns or 'Y-Midpoint' not in data.columns or 'Time' not in data.columns:
                print(f"Required columns not found in {gui_corrected_file}. Skipping.")
                continue

            # Remove the first and last 60 frames
            if len(data) > 120:
                data = data.iloc[60:-60]
            else:
                print(f"Not enough frames in {gui_corrected_file}. Skipping.")
                continue

            # Smooth the data using Savitzky-Golay filter
            window_length = 31  # Must be odd and less than the length of the data
            polyorder = 3       # Polynomial order for the filter
            data['X-Midpoint'] = savgol_filter(data['X-Midpoint'], window_length, polyorder)
            data['Y-Midpoint'] = savgol_filter(data['Y-Midpoint'], window_length, polyorder)

            # Calculate the speed
            x_midpoint = data['X-Midpoint'].values
            y_midpoint = data['Y-Midpoint'].values
            time = data['Time'].values

            # Calculate the differences in positions and time
            dx = np.diff(x_midpoint)
            dy = np.diff(y_midpoint)
            dt = np.diff(time)

            # Avoid division by zero
            dt[dt == 0] = np.nan

            # Calculate speed (pixels per unit time)
            speed = np.sqrt(dx**2 + dy**2) / dt
            speed = np.clip(speed, 0, np.nanpercentile(speed, 95))  # Remove extreme outliers

            # Append the average speed for this experiment
            all_speeds.append(np.nanmean(speed))

        return all_speeds

    # Calculate speeds for both folders
    speeds1 = calculate_speeds(folder1)
    speeds2 = calculate_speeds(folder2)

    # Display the results using box plots
    plt.figure(figsize=(10, 6))
    plt.boxplot([speeds1, speeds2], labels=['Baseline 1', 'Baseline 2'])
    plt.ylabel("Speed (pixels/s)")
    plt.title("Comparison of Speeds Between Two Baseline Folders")
    plt.grid(True)
    plt.show()

    # Print summary statistics
    print(f"Baseline 1 - Average Speed: {np.nanmean(speeds1):.2f} pixels/s, Median: {np.nanmedian(speeds1):.2f} pixels/s")
    print(f"Baseline 2 - Average Speed: {np.nanmean(speeds2):.2f} pixels/s, Median: {np.nanmedian(speeds2):.2f} pixels/s")

# Example usage
folder1 = '/Users/eneko/Desktop/Louis Lab/data collection/251002/b1'  # Replace with your folder path
folder2 = '/Users/eneko/Desktop/Louis Lab/data collection/251002/b2'  # Replace with your folder path
compare_baseline_speeds_with_boxplots(folder1, folder2)
```

```python
import pandas as pd
import numpy as np
from scipy.ndimage import gaussian_filter

import matplotlib.pyplot as plt

def plot_angle_with_stimulation(base_folder, stimulation_column="stimulation", stim_on_value=80000.0, sigma=10):
    """
    Plot the angle between the tail-to-midpoint vector and the midpoint-to-head vector as a function of time,
    with the stimulus overlaid.

    Parameters:
        base_folder (str): Path to the folder containing subfolders with GUI-corrected files.
        stimulation_column (str): The column name representing stimulation (e.g., "stimulation" or "Stim Ch1").
        stim_on_value (float): The value in the stimulation column that indicates stimulation is on.
        sigma (int): The standard deviation for Gaussian smoothing.

    Returns:
        None
    """
    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv") and not f.startswith("aligned_")
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Tail', 'Y-Tail', 'X-Midpoint', 'Y-Midpoint', 'X-Head', 'Y-Head', stimulation_column, 'Time']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Extract coordinates and stimulation data
        x_tail, y_tail = data['X-Tail'].values, data['Y-Tail'].values
        x_mid, y_mid = data['X-Midpoint'].values, data['Y-Midpoint'].values
        x_head, y_head = data['X-Head'].values, data['Y-Head'].values
        stimulation = data[stimulation_column].values
        time = data['Time'].values

        # Calculate vectors
        tail_to_mid = np.array([x_mid - x_tail, y_mid - y_tail]).T
        mid_to_head = np.array([x_head - x_mid, y_head - y_mid]).T

        # Calculate angles between vectors
        dot_product = np.einsum('ij,ij->i', tail_to_mid, mid_to_head)
        norm_tail_to_mid = np.linalg.norm(tail_to_mid, axis=1)
        norm_mid_to_head = np.linalg.norm(mid_to_head, axis=1)
        cos_theta = dot_product / (norm_tail_to_mid * norm_mid_to_head)
        angles = np.arccos(np.clip(cos_theta, -1.0, 1.0))  # Clip to avoid numerical errors

        # Smooth the angles using a Gaussian filter
        angles_smoothed = gaussian_filter(angles, sigma=sigma)

        # Plot the angles as a function of time
        plt.figure(figsize=(10, 6))
        plt.plot(time, angles_smoothed, label="Angle (radians)", color="blue")

        # Overlay stimulation
        stimulation_on = stimulation == stim_on_value
        for i in range(len(stimulation_on) - 1):
            if stimulation_on[i] and stimulation_on[i + 1]:  # Continuous "on" section
                plt.axvspan(time[i], time[i + 1], color="orange", alpha=0.3, label="Stimulation On" if i == 0 else "")

        plt.xlabel("Time (s)")
        plt.ylabel("Angle (radians)")
        plt.title(f"Angle Between Vectors and Stimulation for {folder}")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/260408/1s_onoff_dropout' # Replace with your folder path
plot_angle_with_stimulation(base_folder, stimulation_column="Stim Ch1", stim_on_value=80.0, sigma=10)
```

```python
import os
import pandas as pd
import numpy as np
from scipy.ndimage import gaussian_filter
import matplotlib.pyplot as plt

def calculate_average_angle(base_folder, stimulation_column="stimulation", stim_on_value=80000.0, sigma=10):
    """
    Calculate and plot the average angle between vectors across multiple trials with stimulation overlay.

    Parameters:
        base_folder (str): Path to the folder containing subfolders with GUI-corrected files.
        stimulation_column (str): The column name representing stimulation (e.g., "stimulation" or "Stim Ch1").
        stim_on_value (float): The value in the stimulation column that indicates stimulation is on.
        sigma (int): The standard deviation for Gaussian smoothing.

    Returns:
        None
    """
    all_angles = []
    stim_on_periods = []
    common_time_grid = None  # Define a common time grid for alignment

    for folder in os.listdir(base_folder):
        folder_path = os.path.join(base_folder, folder)
        if not os.path.isdir(folder_path):
            continue

        # Find the GUI-corrected file in the folder
        gui_corrected_files = [
            f for f in os.listdir(folder_path)
            if f.endswith("GUIcorrected.csv") and not f.startswith("aligned_")
        ]
        if not gui_corrected_files:
            print(f"No GUI-corrected file found in {folder_path}. Skipping.")
            continue

        gui_corrected_file = gui_corrected_files[0]
        gui_corrected_path = os.path.join(folder_path, gui_corrected_file)

        # Read the GUI-corrected file
        data = pd.read_csv(gui_corrected_path)

        # Ensure required columns exist
        required_columns = ['X-Tail', 'Y-Tail', 'X-Midpoint', 'Y-Midpoint', 'X-Head', 'Y-Head', stimulation_column, 'Time']
        if not all(col in data.columns for col in required_columns):
            print(f"Required columns not found in {gui_corrected_file}. Skipping.")
            continue

        # Remove the first and last 60 frames
        if len(data) > 120:
            data = data.iloc[60:-60]
        else:
            print(f"Not enough frames in {gui_corrected_file}. Skipping.")
            continue

        # Extract coordinates, time, and stimulation data
        x_tail, y_tail = data['X-Tail'].values, data['Y-Tail'].values
        x_mid, y_mid = data['X-Midpoint'].values, data['Y-Midpoint'].values
        x_head, y_head = data['X-Head'].values, data['Y-Head'].values
        time = data['Time'].values
        stimulation = data[stimulation_column].values

        # Ensure time values are unique
        if len(time) != len(np.unique(time)):
            print(f"Duplicate time values found in {gui_corrected_file}. Removing duplicates.")
            _, unique_indices = np.unique(time, return_index=True)
            time = time[unique_indices]
            x_tail, y_tail = x_tail[unique_indices], y_tail[unique_indices]
            x_mid, y_mid = x_mid[unique_indices], y_mid[unique_indices]
            x_head, y_head = x_head[unique_indices], y_head[unique_indices]
            stimulation = stimulation[unique_indices]

        # Calculate vectors
        tail_to_mid = np.array([x_mid - x_tail, y_mid - y_tail]).T
        mid_to_head = np.array([x_head - x_mid, y_head - y_mid]).T

        # Calculate angles between vectors
        dot_product = np.einsum('ij,ij->i', tail_to_mid, mid_to_head)
        norm_tail_to_mid = np.linalg.norm(tail_to_mid, axis=1)
        norm_mid_to_head = np.linalg.norm(mid_to_head, axis=1)
        cos_theta = dot_product / (norm_tail_to_mid * norm_mid_to_head)
        angles = np.arccos(np.clip(cos_theta, -1.0, 1.0))  # Clip to avoid numerical errors

        # Smooth the angles using a Gaussian filter
        angles_smoothed = gaussian_filter(angles, sigma=sigma)

        # Define a common time grid if not already defined
        if common_time_grid is None:
            common_time_grid = np.linspace(time.min(), time.max(), num=500)  # Interpolate to 500 points

        # Interpolate angles to the common time grid
        interpolated_angles = np.interp(common_time_grid, time, angles_smoothed)
        all_angles.append(interpolated_angles)

        # Identify stimulation on periods
        stim_on = stimulation == stim_on_value
        for i in range(len(stim_on) - 1):
            if stim_on[i] and not stim_on[i - 1]:  # Start of stimulation
                start_time = time[i]
            if stim_on[i] and not stim_on[i + 1]:  # End of stimulation
                end_time = time[i]
                stim_on_periods.append((start_time, end_time))

    # Calculate the average angle at each time point
    average_angles = np.nanmean(all_angles, axis=0)

    # Plot the average angles as a function of time
    plt.figure(figsize=(10, 6))
    plt.plot(common_time_grid, average_angles, label="Average Angle (radians)", color="blue")

    # Overlay stimulation periods
    for start, end in stim_on_periods:
        plt.axvspan(start, end, color="orange", alpha=0.3, label="Stimulation On" if start == stim_on_periods[0][0] else "")

    plt.xlabel("Time (s)")
    plt.ylabel("Angle (radians)")
    plt.title("Average Angle Across Trials with Stimulation Overlay")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Example usage
base_folder = '/Users/eneko/Desktop/Louis Lab/data collection/260408/3s_onoff_dropout_6min'  # Replace with your folder path
calculate_average_angle(base_folder, stimulation_column="Stim Ch1", stim_on_value=80.0, sigma=10)
```
